```python
x = "global" 
  
def outer(): 
    x = "enclosing" 
  
    def inner(): 
        x = "local" 
        print(x) 
  
    inner() 
  
outer() 
```

```python
def outer():
    x = 10
    def inner():
        nonlocal x
        x = 20
    inner()
    print(x)
outer()
```

```python
def square(x): 
    return x * x 
  
def execute(func, value): 
    return func(value) 
  
print(execute(square, 5)) 
```

Create a function that checks whether a number is even or odd.


```python
def check_even_odd(num):
    if num % 2 == 0:
        return "Even"
    else:
        return "Odd"

print(check_even_odd(7))
```

Create a function that returns the larger number.

```python
def maximum(a, b):
    if a > b:
        return a
    return b

print(maximum(10, 20))
```

Create a function that calculates the area of a rectangle.

```python
def area(length, width):
    return length * width

print(area(5, 3))
```

Create a function that returns the length of a string.

```python
def count_chars(text):
    return len(text)

print(count_chars("Python"))
```
