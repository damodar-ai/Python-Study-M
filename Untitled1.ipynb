```python
def show(*s):
    print(type(s))
    print(s)
show(1 , 2)
```

```python
def add(*a):
    print (sum(a))
add(10,50 , 52)
```

```python
def show(*args):
    print(args[0])
    print(args[1])

show(100, 200, 300)
```

Example:
The * tells Python:
Collect all extra positional arguments into a tuple.

```python
def test(*args):
    print(args)
test(1, 2, 3, 4)
```

```python
#*args with Normal Parameters
def student(name, *marks):
    print("Name:", name)
    print("Marks:", marks)

student("Aman", 80, 90, 85)
```

```python
def show(**k):
    print(k)
    print(type(k))
    print(k["name"])
show(name = "aman" , age = "20")
```

```python
def test(*args, **kwargs):

    print("ARGS:", args)
    print("KWARGS:", kwargs)

test(10, 20, 30, name="Aman", age=20)
```
